### Data Loading

#### Import Libraries

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

#### File Paths

```python
email_path = "../Dataset/email.csv"
psycho_path = "../Dataset/psychometric.csv"
```

#### Load Datasets

```python
email_df = pd.read_csv(email_path)
psycho_df = pd.read_csv(psycho_path)

print("Email Dataset Loaded ✅")
print("Psychometric Dataset Loaded ✅")
```

#### Basic Inspection (Email)

```python
print("EMAIL DATASET")
print("Shape:", email_df.shape)
display(email_df.head())
email_df.info()
```

#### Basic Inspection (Psychometric)

```python
print("PSYCHOMETRIC DATASET")
print("Shape:", psycho_df.shape)
display(psycho_df.head())
psycho_df.info()
```

#### Check Missing Values

```python
print("Missing Values - Email")
print(email_df.isnull().sum())

print("\nMissing Values - Psychometric")
print(psycho_df.isnull().sum())
```

#### Duplicate Check

```python
print("Duplicate rows - Email:", email_df.duplicated().sum())
print("Duplicate rows - Psychometric:", psycho_df.duplicated().sum())
```

```python
print("Email Columns:\n", email_df.columns)
print("\nPsychometric Columns:\n", psycho_df.columns)
```

## Insider Threat Radar – Behavior Analytics for Employee Risk Prediction

### Data Cleaning 

#### Import Libraries

```python
import pandas as pd
import numpy as np
```

#### Load Data

```python
email_df = pd.read_csv("../Dataset/email.csv", low_memory=False)
psycho_df = pd.read_csv("../Dataset/psychometric.csv")

print("Datasets Loaded ✅")
```

#### Check Initial Shape

```python
print("Before Cleaning - Email:", email_df.shape)
print("Before Cleaning - Psychometric:", psycho_df.shape)
```

#### Handle missing values

**Handle Missing CC & BCC**

```python
email_df['cc'] = email_df['cc'].fillna("None")
email_df['bcc'] = email_df['bcc'].fillna("None")
```

#### Verify Missing Values

```python
print(email_df[['cc','bcc']].isnull().sum())
```

#### Convert Date Properly

```python
email_df['date'] = pd.to_datetime(email_df['date'], errors='coerce')
```

#### Sort User-wise Chronologically

```python
email_df = email_df.sort_values(by=['user','date'])
email_df.head()
```

#### Remove Duplicate Rows

```python
email_df.drop_duplicates(inplace=True)
psycho_df.drop_duplicates(inplace=True)

print("After duplicate removal:")
print("Email:", email_df.shape)
print("Psychometric:", psycho_df.shape)
```

#### Remove Unnecessary Columns

```python
email_df_clean = email_df.drop(columns=['content'])
```

#### Reset Index

```python
email_df_clean.reset_index(drop=True, inplace=True)
```

#### Save Clean Data

```python
email_df_clean.to_csv("../Dataset/email_clean.csv", index=False)
psycho_df.to_csv("../Dataset/psychometric_clean.csv", index=False)

print("Cleaned files saved ✅")
```
